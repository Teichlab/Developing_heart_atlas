# Function

```python
def clean_adata(adata,
                obs_keep=None,
                uns_keep=None,
                obsm_keep=None,
                obsp_keep=None):
    """
    Keep only specified keys in AnnData slots.

    Parameters
    ----------
    adata : AnnData
    obs_keep : list
        Columns in adata.obs to keep
    uns_keep : list
        Keys in adata.uns to keep
    obsm_keep : list
        Keys in adata.obsm to keep
    obsp_keep : list
        Keys in adata.obsp to keep
    """

    if obs_keep is not None:
        drop = [c for c in adata.obs.columns if c not in obs_keep]
        adata.obs.drop(columns=drop, inplace=True)

    if uns_keep is not None:
        drop = [k for k in list(adata.uns.keys()) if k not in uns_keep]
        for k in drop:
            del adata.uns[k]

    if obsm_keep is not None:
        drop = [k for k in list(adata.obsm.keys()) if k not in obsm_keep]
        for k in drop:
            del adata.obsm[k]

    if obsp_keep is not None:
        drop = [k for k in list(adata.obsp.keys()) if k not in obsp_keep]
        for k in drop:
            del adata.obsp[k]

    return adata
```
